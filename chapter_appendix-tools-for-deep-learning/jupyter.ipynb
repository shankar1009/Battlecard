```python
from bs4 import BeautifulSoup

# Assuming soup is already defined and contains the page content

divs = soup.find_all('div', class_="_cDEzb_p13n-sc-css-line-clamp-1_1Fn1y")

delimiters = [':', '।', ',', '|', '(', ')']

def clean_title(title):
    positions = [title.find(d) for d in delimiters if title.find(d) != -1]
    if positions:
        split_pos = min(positions)
        return title[:split_pos].strip()
    else:
        return title.strip()

for i in range(0, len(divs), 2):
    book_title_raw = divs[i].get_text(strip=True)
    author = divs[i+1].get_text(strip=True) if (i+1) < len(divs) else "Unknown Author"

    book_title = clean_title(book_title_raw)

    print(f"Book: {book_title}")
    print(f"Author: {author}")
    print('-' * 30)
```
